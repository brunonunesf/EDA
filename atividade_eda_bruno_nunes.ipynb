### Quando transformar categorias em números?

Para tabelas e gráficos, normalmente podemos manter os nomes. Em algumas análises, uma representação numérica pode ser útil:

- uma categoria com **dois valores** pode virar uma coluna 0/1 usando `map()`;
- uma categoria **nominal** com poucos valores pode gerar uma coluna indicadora para cada grupo usando `pd.get_dummies()`;

```python
#analise grafica com a nova variavel em relação às demais
# a ideia que quero fazer é um mapa de calor para ver quem melhor se relaciona com a média
categoricas = ["gender", "race/ethnicity", "parental level of education", "lunch", "test preparation course"]

# converter cada uma em 0 ou 1
df_heatmap = pd.get_dummies(df[categoricas], dtype=int)
df_heatmap["media"] = df["media"]

#tive que pesquisar ✌️✌️✌️

#.         calcula a correlacao so em relacao à media (por isso [["media"]], para ficar no formato de DataFrame)
#.                       |.                remove a linha que compara a media com ela mesma
#.                       |.                  |.            ordena os valores; ascending=False -> ordem decrescente
#.                       |.                  |.               |.
correlacoes = (df_heatmap.corr()[["media"]].drop("media")).sort_values(by="media", ascending=False)
sns.heatmap(correlacoes, annot=True) # -> crie o heatmap; annot=True coloca os valores dentro de cada celula

plt.title("correlações entre as variáveis e a média")
plt.show()
```
